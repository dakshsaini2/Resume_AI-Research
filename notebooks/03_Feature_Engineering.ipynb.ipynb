```python
# Standardize column names
resume_df.columns = resume_df.columns.str.strip().str.lower()

candidate_columns = [
    "career_objective",
    "skills",
    "degree_names",
    "major_field_of_studies",
    "professional_company_names",
    "positions",
    "certification_skills"
]

job_columns = [
    "job_position_name",
    "skills_required",
    "educational_requirements",
    "experiencere_requirement"
]

# Fill missing values
resume_df[candidate_columns + job_columns] = (
    resume_df[candidate_columns + job_columns].fillna("")
)

# Create profiles
resume_df["candidate_profile"] = (
    resume_df[candidate_columns]
    .astype(str)
    .agg(" ".join, axis=1)
)

resume_df["job_profile"] = (
    resume_df[job_columns]
    .astype(str)
    .agg(" ".join, axis=1)
)

# Verify
resume_df[["candidate_profile", "job_profile", "matched_score"]].head()
```
